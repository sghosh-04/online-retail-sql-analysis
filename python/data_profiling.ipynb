Data Quality Assessment

```python
missing = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing %": (df.isnull().mean() * 100).round(2)
})

missing = missing.sort_values("Missing %", ascending=False)

missing.head(20)
```

```python
missing_percentage = missing["Missing %"]

summary = pd.DataFrame({
    "Category": [
        "0%",
        "0-10%",
        "10-30%",
        "30-50%",
        "50-70%",
        "70-90%",
        ">90%"
    ],
    "Columns": [
        (missing_percentage == 0).sum(),
        ((missing_percentage > 0) & (missing_percentage <= 10)).sum(),
        ((missing_percentage > 10) & (missing_percentage <= 30)).sum(),
        ((missing_percentage > 30) & (missing_percentage <= 50)).sum(),
        ((missing_percentage > 50) & (missing_percentage <= 70)).sum(),
        ((missing_percentage > 70) & (missing_percentage <= 90)).sum(),
        (missing_percentage > 90).sum()
    ]
})

summary
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(summary["Category"], summary["Columns"])

plt.title("Missing Value Distribution")
plt.xlabel("Missing Percentage")
plt.ylabel("Number of Columns")

plt.show()
```

```python
missing.to_csv("../reports/missing_value_report.csv")
summary.to_csv("../reports/missing_summary.csv", index=False)
```
