```python
import pandas as pd
import numpy as np

# Display all columns while working
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
```

```python
df = pd.read_csv("../data/raw/GlobalFindexDatabase2025.csv")
```

```python
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])

df.head()
```

```python
dim_country = (
    df[
        [
            "countrynewwb",
            "codewb",
            "regionwb24_hi",
            "incomegroupwb24"
        ]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
dim_country.head()
```

```python
print(dim_country.shape)
```

```python
dim_country.insert(
    0,
    "CountryID",
    range(1, len(dim_country) + 1)
)
```

```python
dim_country.head()
```

```python
print("Duplicate Country IDs :", dim_country["CountryID"].duplicated().sum())
print("Duplicate Country Codes :", dim_country["codewb"].duplicated().sum())
print("Null Values\n")
print(dim_country.isnull().sum())
```

```python
dim_date = (
    df[["year"]]
    .drop_duplicates()
    .sort_values("year")
    .reset_index(drop=True)
)
```

```python
dim_date.insert(
    0,
    "YearID",
    range(1, len(dim_date) + 1)
)
```

```python
dim_date
```

```python
dim_demo = (
    df[
        [
            "group",
            "group2"
        ]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
dim_demo.insert(
    0,
    "GroupID",
    range(1, len(dim_demo) + 1)
)
```

```python
dim_demo
```

```python
fact = df.copy()
```

```python
fact = fact.merge(
    dim_country,
    on=[
        "countrynewwb",
        "codewb",
        "regionwb24_hi",
        "incomegroupwb24"
    ],
    how="left"
)
```

```python
fact[["countrynewwb", "CountryID"]].head()
```

```python
fact = fact.merge(
    dim_date,
    on="year",
    how="left"
)
```

```python
fact[["year", "YearID"]].head()
```

```python
fact = fact.merge(
    dim_demo,
    on=[
        "group",
        "group2"
    ],
    how="left"
)
```

```python
fact[["group", "group2", "GroupID"]].head()
```

```python
print("Missing CountryID :", fact["CountryID"].isnull().sum())
print("Missing YearID    :", fact["YearID"].isnull().sum())
print("Missing GroupID   :", fact["GroupID"].isnull().sum())
```

```python
fact.drop(
    columns=[
        "countrynewwb",
        "codewb",
        "regionwb24_hi",
        "incomegroupwb24",
        "year",
        "group",
        "group2"
    ],
    inplace=True
)
```

```python
key_columns = ["CountryID", "YearID", "GroupID"]

other_columns = [col for col in fact.columns if col not in key_columns]

fact = fact[key_columns + other_columns]
```

```python
print(fact.shape)
fact.head()
```

```python
import os

output_folder = "../data/processed"

os.makedirs(output_folder, exist_ok=True)
```

```python
dim_country.to_csv(
    f"{output_folder}/DimCountry.csv",
    index=False
)
```

```python
dim_date.to_csv(
    f"{output_folder}/DimDate.csv",
    index=False
)
```

```python
dim_demo.to_csv(
    f"{output_folder}/DimDemographic.csv",
    index=False
)
```

```python
fact.to_csv(
    f"{output_folder}/FactFinancialIndicators.csv",
    index=False
)
```
